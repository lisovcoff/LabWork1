## 2. NumPy — арифметические операции над массивами

```python
import numpy as np
mas1 = np.array([34, 45, 56, 0, -1])
mas2 = np.array([2, 5, -3, 10, 4])
print("mas1:", mas1)
print("mas2:", mas2)
print("Сложение:", mas1 + mas2)
print("Вычитание:", mas1 - mas2)
print("Умножение:", mas1 * mas2)
print("Деление:", mas1 / mas2)
print("Целочисленное деление:", mas1 // mas2)
print("Остаток:", mas1 % mas2)
print("Степень:", mas1 ** 2)
```

## 4. Работа аэропорта NNN
Создаём двумерный NumPy-массив из трёх столбцов: вылеты, прилёты и их сумма.

```python
np.random.seed(42)
n = 30
departures = np.random.randint(15, 65, size=n)
arrivals = np.random.randint(15, 65, size=n)
airport = np.column_stack((departures, arrivals, departures + arrivals))
print("Размер массива:", airport.shape)
print("Среднее вылетов:", f"{airport[:,0].mean():.2f}")
print("Среднее прилётов:", f"{airport[:,1].mean():.2f}")
print("Среднее всего рейсов:", f"{airport[:,2].mean():.2f}")
half1 = airport[:15, 2].sum()
half2 = airport[15:, 2].sum()
print("Первая половина:", half1)
print("Вторая половина:", half2)
print("Больше рейсов: первая половина" if half1 > half2 else "Больше рейсов: вторая половина" if half2 > half1 else "Половины равны")
saturday = airport[5::7, 1]
print("Субботы по прилётам:", saturday)
print("Вылетов > 10:", np.sum(airport[:,0] > 10))
print("10 < вылеты < 50:", np.sum((airport[:,0] > 10) & (airport[:,0] < 50)))
```

## 5. Операции над матрицами — `linalg`
Для матрицы `A` найдём чётные элементы, `A·Aᵀ`, обратную матрицу и проверим произведение с обратной матрицей.

```python
np.random.seed(42)
A = np.random.randint(1, 11, size=(3, 3))
even_elements = A[A % 2 == 0]
AAT = A @ A.T
AAT_inv = np.linalg.inv(AAT)
B = AAT @ AAT_inv
print("Чётные элементы:", even_elements)
print("Сумма чётных:", even_elements.sum())
print("det(B):", round(np.linalg.det(B), 4))
```
